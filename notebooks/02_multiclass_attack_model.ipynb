# Multiclass Attack Classification

Goal: train a model to classify different types of network traffic, not only BENIGN vs DDoS.

```python
# 1. Load datasets 
import pandas as pd
from pathlib import Path

data_path= Path("../data")
csv_files= list(data_path.glob("*.csv"))
```

```python
# 2. Combine datasets
dfs= []
for file in csv_files:
    df_tmp= pd.read_csv(file)
    dfs.append(df_tmp)

df= pd.concat(dfs, ignore_index=True)
df.columns= df.columns.str.strip()

df.shape
```

```python
# 3. Analyze labels frequency and clean some data

df["Label"] = df["Label"].str.replace("�", "-", regex=False)

print(df["Label"].value_counts())
df["Label"].value_counts(normalize=True)
```

```python
## 4. Remove extremely rare classes
'''
Some attack classes have very few examples in the dataset, such as Heartbleed, Infiltration and SQL Injection. 
These classes were removed at this stage because there are too few samples for the model to learn reliable patterns.
The XSS class was kept because it has more examples and is still useful for multiclass classification.
'''
classes_to_remove= ["Heartbleed", "Infiltration", "Web Attack - Sql Injection"]
for label in classes_to_remove:
    df=df[df["Label"]!= label]
```

```python
## 5. Encoding the labels

from sklearn.preprocessing import LabelEncoder

X= df.drop(columns=['Label'])
y=df['Label']
encoder= LabelEncoder()
y_encoded = encoder.fit_transform(y)

#labels encoded print
for class_id, class_name in enumerate(encoder.classes_):
    print(class_id, "->", class_name)
```
